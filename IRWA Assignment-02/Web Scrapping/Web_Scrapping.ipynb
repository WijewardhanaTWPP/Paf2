# Scrap User Reviews about Given Movie

```python
url=f"https://www.imdb.com/title/tt{movie_id}/reviews?ref_=tt_ov_rt"

response=requests.get(url)

all_user_reviews=" "

if(response.status_code==200):

  soup=BeautifulSoup(response.content,"lxml")

  all_user_reviews=soup.find_all("div",{"class":"text show-more__control"})

else:
  print("Cannot Scrap the Given URL!!!")
```

```python
reviews=[review.get_text() for review in all_user_reviews]
```

```python
print(len(reviews))
```

# Reviews Classification

```python
for r in reviews:
  print(r)
```

```python
# import Count Vectorizer
count_vectorizer=pickle.load(open("/content/drive/MyDrive/IRWA Assignment-02/Review Classification/Count Vectorizer.pkl", 'rb'))
```

```python
# import Review Classification Model
review_classification_model=pickle.load(open("/content/drive/MyDrive/IRWA Assignment-02/Review Classification/Review Classification Model.pkl", 'rb'))
```

```python
#review classification model
review_status=dict()

for review in reviews:
  cv=count_vectorizer.transform([review])

  prediction=review_classification_model.predict(cv)[0]

  status=""
  if(prediction==1):
      status+="Good"
  else:
      status+="Bad"


  review_status[review]=status
```

```python
print(review_status)
```

```python
for r,s in review_status.items():
  print(f"Review: {r} =====> {s}")
```
